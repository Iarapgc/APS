## 2. Introducción teórica

### 2.1 Filtros analógicos y función transferencia

La **función transferencia** es una herramienta matemática que describe el comportamiento de un sistema lineal e invariante en el tiempo (LTI):

$$H(s)=\frac{V_{salida}}{V_{entrada}}=\frac{Y(s)}{X(s)}\qquad \text{con } s=\sigma+j\omega$$

### 2.2 Respuesta en frecuencia

La respuesta en frecuencia describe cómo reacciona el sistema cuando se le dan señales de distintas frecuencias. Se obtiene evaluando la función transferencia sobre el eje imaginario, $s=j\omega$:

$$H(j\omega)=H(s)\big|_{s=j\omega}\;\longrightarrow\;|H(\omega)|,\quad \angle H(\omega)$$

Esto da el módulo $|H(\omega)|$ y la fase $\angle H(\omega)$ del filtro en función de la frecuencia $\omega$.

$$|H|=\sqrt{\text{Re}^2+\text{Im}^2}\qquad\qquad \angle H=\operatorname{arctg}\!\left(\frac{\text{Im}}{\text{Re}}\right)=\sum\angle\text{ceros}-\sum\angle\text{polos}$$



**Estabilidad:** para que el sistema sea estable, todos los polos deben estar en el semiplano izquierdo, es decir, $\text{Re}\{s_p\}<0$.

### 2.3 Polos, ceros y orden

- Los **ceros** son los valores de $s$ donde $H(s)=0$, es decir, donde $V_{salida}=Y(s)=0$ (raíces del numerador).
- Los **polos** son los valores de $s$ donde $H(s)\to\infty$ (raíces del denominador).
- El **orden** está dado por el grado del polinomio del denominador. En circuitos pasivos coincide con la cantidad de elementos reactivos que almacenan energía ($L$ y $C$): si hay $L$ y $C$ es de orden 2; si solo hay uno de ellos, es de orden 1.

### 2.4 Tipos de filtros

| Filtro | $H(s)$ | Valores límite | Comportamiento | Rasgo en $H(s)$ |
|:--|:--:|:--|:--|:--|
| **Pasa-bajos** | $\dfrac{K}{1+s\tau}$ | $H(0)=K\neq0$, $H(\infty)=0$ | Deja pasar bajas frecuencias y atenúa las altas | Numerador constante, denominador con $s$ |
| **Pasa-altos** | $\dfrac{s\tau}{1+s\tau}$ | $H(0)=0$, $H(\infty)=K\neq0$ | Deja pasar altas frecuencias y bloquea las bajas | Numerador con $s$; se anula en $s=0$ |
| **Pasa-banda** | $\dfrac{s}{s^2+as+b}$, con $a=\frac{R}{L}$, $b=\frac{1}{LC}$ | $H(0)=0$, $H(\infty)=0$ | Solo deja pasar un rango intermedio | Numerador con $s$, denominador de 2.º orden |
| **Rechaza-banda (notch)** | $\dfrac{s^2+\omega_0^2}{s^2+as+b}$ | $H(0)\neq0$, $H(\infty)\neq0$ | Se anula a una frecuencia intermedia; rechaza una banda específica | Ceros sobre el eje $j\omega$ |

### 2.5 Forma canónica de segundo orden y relaciones útiles

Todos los circuitos de esta tarea tienen el mismo denominador:

$$s^2+\frac{\omega_0}{Q}\,s+\omega_0^2$$

con $\omega_0=\dfrac{1}{\sqrt{LC}}$ la frecuencia natural y $Q$ el factor de calidad. Con $Q=\frac{\sqrt2}{2}$ los polos son:

$$s_{1,2}=-\frac{\omega_0}{2Q}\pm j\,\omega_0\sqrt{1-\frac{1}{4Q^2}}=-\frac{\sqrt2}{2}\pm j\frac{\sqrt2}{2}=e^{\pm j\,3\pi/4}$$

Es decir, dos polos complejos conjugados sobre la circunferencia de radio $\omega_0=1$, con un ripple $\zeta=\frac{1}{2Q}=\frac{\sqrt2}{2}$. Para el pasa-bajos esto corresponde a una respuesta **Butterworth** (máximamente plana).

Para los pasa-banda, el ancho de banda a $-3\,\text{dB}$ cumple $BW=\dfrac{\omega_0}{Q}$.

## 3. Primer circuito (RL serie + C en paralelo a la salida)

WhatsApp-Image-2026-10-11-at-11.04.41-PM.png

### 3.1 Análisis preliminar

La impedancia serie es $Z_1=sL+R$ y la de la salida es $Z_2=Z_C=\dfrac{1}{sC}$.

- En $\omega=0$: $Z_L=0$ y $Z_C=\infty$ → todo $V_1$ aparece en la salida.
- En $\omega\to\infty$: $Z_L=\infty$ y $Z_C=0$ → la salida se cortocircuita.

Entonces se trata de un **filtro pasa-bajos** (de 2.º orden, porque hay un $L$ y un $C$).

### 3.2 Función de transferencia

$$H(s)=\frac{V_2}{V_1}=\frac{Z_2}{Z_1+Z_2}=\frac{\frac{1}{sC}}{sL+R+\frac{1}{sC}}=\frac{1}{s^2LC+sRC+1}=\frac{\frac{1}{LC}}{s^2+\frac{R}{L}s+\frac{1}{LC}}$$

**Normalización.** Comparando con $\dfrac{\omega_0^2}{s^2+\frac{\omega_0}{Q}s+\omega_0^2}$:

$$\omega_0^2=\frac{1}{LC}=1,\qquad \frac{\omega_0}{Q}=\frac{R}{L}=\sqrt2,\qquad R=1\ \Omega$$

$$\Rightarrow\; L=\frac{RQ}{\omega_0}=\frac{\sqrt2}{2}\ \approx0{,}707\ \qquad C=\frac{1}{\omega_0^2L}=\sqrt2\ \approx1{,}414\ $$

$$\boxed{H(s)=\frac{1}{s^2+\sqrt2\,s+1}}$$

### 3.3 Polos y ceros

- **Ceros:** no hay ceros finitos; ambos están "implícitos" en $s\to\infty$ (grado del denominador − grado del numerador = 2).
- **Polos:**

$$s_1=-\frac{\sqrt2}{2}+j\frac{\sqrt2}{2},\qquad s_2=-\frac{\sqrt2}{2}-j\frac{\sqrt2}{2}$$

Ambos con $\text{Re}<0$ → el sistema es estable.

WhatsApp-Image-2026-10-11-at-11.07.33-PM.png

### 3.4 Módulo y fase

Con $s=j\omega$:

$$H(j\omega)=\frac{1}{1-\omega^2+j\sqrt2\,\omega}$$

$$\boxed{|H(\omega)|=\frac{1}{\sqrt{(1-\omega^2)^2+2\omega^2}}=\frac{1}{\sqrt{1+\omega^4}}}$$

$$\boxed{\angle H(\omega)=-\operatorname{arctg}\!\left(\frac{\sqrt2\,\omega}{1-\omega^2}\right)}$$

**Puntos característicos:**

| $\omega$ | $\omega\to0$ | $\omega=1$ | $\omega\to\infty$ |
|:--:|:--:|:--:|:--:|
| $\lvert H\rvert$ | $1$ (0 dB) | $\frac{1}{\sqrt2}$ (−3 dB) | $\to0$ |
| $\angle H$ | $0$ | $-\frac{\pi}{2}$ | $-\pi$ |

**Respuesta asintótica:**

- Módulo: $0\ \text{dB}$ para $\omega<1$ y pendiente de $-40\ \text{dB/dec}$ para $\omega>1$: $|H|_{dB}\approx-40\log_{10}\omega$.
- Fase: $0$ para bajas frecuencias y $-\pi$ para altas frecuencias; la transición pasa por $-\pi/2$ en $\omega_0$.

### 3.5 Tipo de filtro y ancho de banda

- **Tipo:** filtro **pasa-bajos de 2.º orden** con respuesta **Butterworth** ($Q=1/\sqrt2$): no tiene sobrepico y es máximamente plano. En efecto, $|H|^2=\frac{1}{1+\omega^4}$.
- **Ancho de banda:** la frecuencia de corte a $-3\,\text{dB}$ es $|H|=\frac{1}{\sqrt2}\Rightarrow\omega_c=\omega_0=1\ \text{rad/s}$. Por lo tanto $BW=\omega_c-0=\mathbf{1\ rad/s}$ (equivale a $f_c=\frac{1}{2\pi}\approx0{,}159$ Hz).

## 4. Segundo circuito (C en paralelo con R en serie + L)

C2.jpeg

### 4.1 Análisis preliminar

La impedancia serie es $Z_i=Z_C\parallel Z_R$ y la de la salida es $Z_o=Z_L=sL$.

- En $\omega=0$: $Z_L=0$ y $Z_C=\infty$ → la salida queda en cortocircuito, $V_2=0$.
- En $\omega\to\infty$: $Z_L=\infty$ y $Z_C=0$ → $Z_i\to0$ y $V_2=V_1$.

Entonces se trata de un **filtro pasa-altos**.

### 4.2 Función de transferencia

$$Z_i=Z_C\parallel Z_R=\frac{\frac{R}{sC}}{\frac1{sC}+R}=\frac{R}{1+sRC},\qquad Z_o=sL$$

$$H(s)=\frac{Z_o}{Z_i+Z_o}=\frac{sL}{\frac{R}{1+sRC}+sL}=\frac{sL\,(1+sRC)}{R+sL+s^2RLC}=\frac{RLC\left(s^2+\frac{s}{RC}\right)}{RLC\left(s^2+\frac{s}{RC}+\frac1{LC}\right)}$$

$$H(s)=\frac{s\left(s+\frac1{RC}\right)}{s^2+\frac{1}{RC}s+\frac1{LC}}$$

**Normalización** ($\omega_0=1$, $R=1\ \Omega$, $Q=\frac{\sqrt2}{2}$):

$$\frac1{LC}=\omega_0^2=1,\qquad \frac1{RC}=\frac{\omega_0}{Q}=\sqrt2\;\Rightarrow\; C=\frac{\sqrt2}{2}\ \text{F},\quad L=\frac{1}{\omega_0^2C}=\sqrt2\ \text{H}$$

$$\boxed{H(s)=\frac{s\,(s+\sqrt2)}{s^2+\sqrt2\,s+1}}$$

### 4.3 Polos y ceros

- **Ceros:** $s=0$ y $s=-\dfrac{2}{\sqrt2}=-\sqrt2$.
- **Polos:** $s_{1,2}=-\dfrac{\sqrt2}{2}\pm j\dfrac{\sqrt2}{2}$ (los mismos que en el circuito 1).

PyC_C2_1.png

### 4.4 Módulo y fase

$$H(j\omega)=\frac{j\omega\,(j\omega+\sqrt2)}{1-\omega^2+j\sqrt2\,\omega}=\frac{-\omega^2+j\sqrt2\,\omega}{1-\omega^2+j\sqrt2\,\omega}$$

$$\boxed{|H(\omega)|=\frac{\omega\sqrt{\omega^2+2}}{\sqrt{(1-\omega^2)^2+2\omega^2}}=\sqrt{\frac{\omega^4+2\omega^2}{1+\omega^4}}}$$

$$\boxed{\angle H(\omega)=\underbrace{\frac{\pi}{2}}_{\text{cero en }0}+\underbrace{\operatorname{arctg}\!\left(\frac{\omega}{\sqrt2}\right)}_{\text{cero en }-\sqrt2}-\underbrace{\operatorname{arctg}\!\left(\frac{\sqrt2\,\omega}{1-\omega^2}\right)}_{\text{polos}}}\qquad$$

**Puntos característicos:**

| $\omega$ | $\omega\to0$ | $\omega=1$ | $\omega\approx1{,}27$ | $\omega\to\infty$ |
|:--:|:--:|:--:|:--:|:--:|
| $\lvert H\rvert$ | $\to0$ ($\approx\sqrt2\,\omega$) | $\sqrt{3/2}\approx1{,}225$ (1,76 dB) | máx. $\approx1{,}272$ (2,09 dB) | $1$ (0 dB) |
| $\angle H$ | $+\frac\pi2$ | $\operatorname{arctg}\frac{1}{\sqrt2}\approx35{,}3^\circ$ | — | $0$ |

El máximo de $|H|^2=\frac{x^2+2x}{1+x^2}$ (con $x=\omega^2$) se da en $x^2-x-1=0\Rightarrow\omega_{max}^2=\frac{1+\sqrt5}{2}$, o sea $\omega_{max}\approx1{,}272$ rad/s: hay un **pequeño sobrepico** de $\approx2$ dB.

## 5. Tercer circuito (R serie + L en paralelo con C a la salida)

C3.jpeg

### 5.1 Análisis preliminar

La impedancia serie es $Z_1=R$ y la de la salida es $Z_2=Z_L\parallel Z_C$.

- En $\omega=0$: $Z_L=0$ y $Z_C=\infty$ → la inductancia cortocircuita la salida, $V_2=0$.
- En $\omega\to\infty$: $Z_L=\infty$ y $Z_C=0$ → el capacitor cortocircuita la salida, $V_2=0$.

Entonces se trata de un **filtro pasa-banda**.

### 5.2 Función de transferencia

$$Z_2=Z_L\parallel Z_C=\frac{sL\cdot\frac1{sC}}{sL+\frac1{sC}}=\frac{\frac{L}{C}}{\frac{1}{sC}+sL}=\frac{sL}{1+s^2LC}=\frac{\frac{s}{C}}{s^2+\frac1{LC}}$$

$$H(s)=\frac{Z_2}{Z_2+Z_1}=\frac{\frac{sL}{1+s^2LC}}{\frac{sL}{1+s^2LC}+R}=\frac{sL}{sL+R+s^2RLC}=\frac{\frac{s}{RC}}{s^2+\frac{s}{RC}+\frac1{LC}}$$

**Normalización** ($\omega_0=1$, $R=1\ \Omega$, $Q=\frac{\sqrt2}{2}$):

$$\frac{1}{LC}=\omega_0^2=1,\qquad \frac{\omega_0}{Q}=\frac{1}{RC}=\sqrt2\;\Rightarrow\; C=\frac{\sqrt2}{2}\ \text{F},\quad L=\frac{1}{\omega_0^2 C}=\sqrt2\ \text{H}$$

$$\boxed{H(s)=\frac{\sqrt2\,s}{s^2+\sqrt2\,s+1}}$$

### 5.3 Polos y ceros

- **Ceros:** $s=0$ y uno en $s\to\infty$.
- **Polos:** $s_{1,2}=-\dfrac{\sqrt2}{2}\pm j\dfrac{\sqrt2}{2}$.

PyC_C3.jpeg

### 5.4 Módulo y fase

$$H(j\omega)=\frac{j\sqrt2\,\omega}{1-\omega^2+j\sqrt2\,\omega}$$

$$\boxed{|H(\omega)|=\frac{\sqrt2\,\omega}{\sqrt{(1-\omega^2)^2+2\omega^2}}=\frac{\sqrt2\,\omega}{\sqrt{1+\omega^4}}}$$

$$\boxed{\angle H(\omega)=\frac{\pi}{2}-\operatorname{arctg}\!\left(\frac{1-\omega^2}{\sqrt2\,\omega}\right)}$$


**Puntos característicos:**

| $\omega$ | $\omega\to0$ | $\omega=1=\omega_0$ | $\omega\to\infty$ |
|:--:|:--:|:--:|:--:|
| $\lvert H\rvert$ | $\to0$ ($\approx\sqrt2\,\omega$) | $1$ (0 dB, **máximo**) | $\to0$ ($\approx\sqrt2/\omega$) |
| $\angle H$ | $+\frac\pi2$ | $0$ | $-\frac\pi2$ |

## 6. Bonus

### 6.1 Cuarto circuito (circuito general con parámetros $a$ y $b$)

CB.jpeg


El circuito tiene dos ramas:

- **Rama admitancia $Y_1$:** $\dfrac{L}{b}\parallel C\,a$, $Y_1=\dfrac{b}{sL}+s\,aC$.
- **Rama admitancia $Y_2$:** $C(1-a)\parallel\dfrac{L}{1-b}\parallel R$, $Y_2=s(1-a)C+\dfrac{1-b}{sL}+\dfrac{1}{R}$.

Como es un divisor de tensión, $H=\dfrac{Z_p}{Z_s+Z_p}=\dfrac{Y_s}{Y_s+Y_p}$, y las partes de $L$ y $C$ se reparten en $a$ y $b$ de modo que la suma total es siempre $L$ y $C$:

$$Y_s+Y_p=\frac{1}{sL}+sC+\frac1R\;\Rightarrow\;H(s)=\frac{\frac{b}{sL}+saC}{\frac{1}{sL}+sC+\frac1R}=\frac{a\,LC\,s^2+b}{LC\,s^2+\frac{L}{R}s+1}$$

$$\boxed{H(s)=\frac{a\,s^2+b\,\omega_0^2}{s^2+\frac{\omega_0}{Q}\,s+\omega_0^2}}\qquad\text{con }\omega_0^2=\frac{1}{LC},\quad \frac{\omega_0}{Q}=\frac{1}{RC}$$

El denominador es siempre el mismo (los polos no dependen de $a$ ni de $b$); los parámetros solo mueven los ceros, en $s=\pm j\,\omega_0\sqrt{b/a}$. Además $H(0)=b$ y $H(\infty)=a$.

| $a$ | $b$ | Tipo de filtro |
|:--:|:--:|:--|
| $0$ | $1$ | Pasa-bajos (como el circuito 1) |
| $1$ | $0$ | Pasa-altos |
| $a=b$ | | **Notch (rechaza-banda)** simétrico, $H(0)=H(\infty)=a$, con cero en $\omega_0$ |
| $b>a$ | | Pasa-bajos con notch en $\omega_0\sqrt{b/a}>\omega_0$ |
| $a>b$ | | Pasa-altos con notch en $\omega_0\sqrt{b/a}<\omega_0$ |

#### Caso notch: $a=b=1$

Con $a=b=1$ desaparecen $C(1-a)$ y $L/(1-b)$ (queda a masa solo $R$), y la rama serie es $L\parallel C$ completa. Con los valores normalizados ($\omega_0=1$, $R=1\ \Omega$, $Q=\frac{\sqrt2}{2}$) resulta $C=\frac{\sqrt2}{2}$ F y $L=\sqrt2$ H:

$$\boxed{H(s)=\frac{s^2+1}{s^2+\sqrt2\,s+1}}$$

- **Ceros:** $s=\pm j$ (sobre el eje imaginario, en $\omega_0$). **Polos:** $-\frac{\sqrt2}{2}\pm j\frac{\sqrt2}{2}$.
- **Módulo:** $|H(\omega)|=\dfrac{|1-\omega^2|}{\sqrt{1+\omega^4}}$.
- **Fase:** $\angle H=-\operatorname{arctg}\!\left(\dfrac{\sqrt2\,\omega}{1-\omega^2}\right)$ para todo $\omega\neq1$. Vale $0$ en ambos extremos ($\omega\to0$ y $\omega\to\infty$), y en $\omega_0=1$, donde $H=0$, tiene un salto de $180^\circ$ (de $-90^\circ$ a $+90^\circ$).
- **Tipo:** **rechaza-banda (notch)** de 2.º orden, con $H(0)=H(\infty)=1$.
- **Ancho de banda de rechazo** ($-3$ dB): $|1-\omega^2|=\sqrt2\,\omega\Rightarrow\omega_{1,2}=\sqrt{\tfrac32}\mp\tfrac{\sqrt2}{2}$, es decir $\omega_1\approx0{,}518$, $\omega_2\approx1{,}932$ y $BW=\sqrt2=\frac{\omega_0}{Q}$ rad/s.

### 6.2 Re-escalado de componentes para $f_0=10\ \text{kHz}$ y $R=50\ \Omega$

**Planteo.** Se busca que la frecuencia natural sea $f_0 = 10\ \text{kHz}$, es decir:

$$\omega_0 = 2\pi f_0 = 2\pi \cdot 10\ \text{kHz} \approx 62\,832\ \text{rad/s}$$

Del denominador de $H(s)$, $\;s^2+\frac{1}{RC}s+\frac{1}{LC}$, y comparándolo con la forma canónica $s^2+\frac{\omega_0}{Q}s+\omega_0^2$, se obtienen dos condiciones:

$$\frac{1}{LC}=\omega_0^2 \qquad\text{y}\qquad \frac{1}{RC}=\frac{\omega_0}{Q}$$

**Cálculo de $C$.** Se elevan al cuadrado ambos miembros de la segunda condición y se usa $Q=\frac{\sqrt2}{2}\Rightarrow\frac{1}{Q^2}=2$:

$$\frac{1}{(RC)^2}=\frac{\omega_0^2}{Q^2}=2\,\omega_0^2=(2\pi\cdot10\ \text{kHz})^2\cdot 2$$

Despejando $C^2$:

$$C^2=\frac{1}{(2\pi\cdot10\ \text{kHz})^2\cdot 2\cdot R^2}\;\Rightarrow\; C=\frac{1}{2\pi\cdot10\ \text{kHz}\cdot\sqrt2\cdot R}$$

$$C=\frac{1}{(2\pi\cdot 10\ \text{kHz})\,\sqrt2\,(50\ \Omega)}\approx 2{,}25\times10^{-7}\ \text{F}=225\ \text{nF}$$

**Cálculo de $L$.** De la primera condición, $\frac{1}{LC}=\omega_0^2$:

$$L=\frac{1}{(2\pi\cdot10\ \text{kHz})^2\cdot C}\approx 1{,}12\times10^{-3}\ \text{H}=1{,}12\ \text{mH}$$

**Resultado:**

$$\boxed{R=50\ \Omega,\qquad C\approx225\ \text{nF},\qquad L\approx1{,}12\ \text{mH}}$$

**En palabras.** Primero se pasa la frecuencia pedida a radianes por segundo. Después se usan las dos relaciones que fijan el comportamiento del circuito: una vincula $L$ y $C$ con la frecuencia natural, y la otra vincula $R$ y $C$ con el factor de calidad $Q$. Como $R=50\ \Omega$ es dato y $Q=\frac{\sqrt2}{2}$, la segunda relación permite despejar $C$. Con ese $C$, la primera relación da $L$.

> Estas relaciones valen para los circuitos 2, 3 y Bonus, donde $\frac{1}{RC}=\frac{\omega_0}{Q}$. En el circuito 1 (pasa-bajos) la condición es $\frac{R}{L}=\frac{\omega_0}{Q}$, por lo que da $L\approx562{,}7\ \mu\text{H}$ y $C\approx450\ \text{nF}$.

## 7. Conclusión

- Los cuatro circuitos tienen la misma forma canónica de segundo orden, por eso tienen los mismos polos, $-\dfrac{\sqrt2}{2}\pm j\dfrac {\sqrt2}{2}$, resultan ubicados en el semiplano izquierdo y son todos estables. La siguiente tabla muestra la función transferencia, el tipo de filtro, las frecuencias de corte y el ancho de banda para cada circuito:

| Circuito | $H(s)$ | Tipo | Frecuencia(s) de corte | Ancho de banda |
|:--|:--:|:--|:--:|:--:|
| 1 | $\dfrac{1}{s^2+\sqrt2 s+1}$ | Pasa-bajos (Butterworth) | $\omega_c=1$ | $1$ rad/s |
| 2 | $\dfrac{s(s+\sqrt2)}{s^2+\sqrt2 s+1}$ | Pasa-altos (leve sobrepico de 2 dB) | $\omega_c\approx0{,}486$ | $\infty$ (teórico) |
| 3 | $\dfrac{\sqrt2\,s}{s^2+\sqrt2 s+1}$ | Pasa-banda | $\omega_1\approx0{,}518$; $\omega_2\approx1{,}932$ | $\sqrt2$ rad/s |
| Bonus ($a=b=1$) | $\dfrac{s^2+1}{s^2+\sqrt2 s+1}$ | Rechaza-banda (notch) | $\omega_1\approx0{,}518$; $\omega_2\approx1{,}932$ | $\sqrt2$ rad/s (rechazo) |

<br>En la columna que indica $H(s)$ se ve cómo solo cambia el numerador, lo cual permite definir el tipo de filtro para cada caso.<br>Con $Q = \dfrac{\sqrt{2}}{2}$, el **pasa-bajos** es Butterworth, por lo que la banda de paso resulta lo más plana posible. El **pasa-altos** tiene un leve sobrepico, consecuencia del cero en $-\sqrt{2}$.<br>
Las dos últimas filas tienen el mismo ancho de banda y las mismas frecuencias de corte. La diferencia es que en el **pasa-banda** esa banda es la que pasa y en el **rechaza-banda** (eligiendo $a=b$) es la que se rechaza. Al tener un Q bajo, ambos resultan poco selectivos.<br>
El circuito **bonus** generaliza casi todos los filtros básicos, su función transferencia mantiene el mismo denominador para cualquier $a$ y $b$, y estos fijan los valores extremos $H(0)=b$ y $H(∞)=a$. <br>Así que:


*   $a=0, b=1$ se obtiene un *pasa-bajos*
*   $a=1, b=0$ se obtiene un *pasa-altos*
*   $a=b$ se obtiene un *rechaza-banda*

Como el numerador no tiene término en $s$, no se puede obtener un *pasa-banda*.

En conclusión, con los mismos polos y el mismo $Q$, la posición de los ceros es lo que define si el circuito deja pasar las bajas frecuencias, las altas, una banda intermedia o si rechaza una banda.
