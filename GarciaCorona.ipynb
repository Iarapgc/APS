```python
# -*- coding: utf-8 -*-
"""
Created on Wed Aug 12 21:15:35 2026

@author: User
"""

import numpy as np
import matplotlib.pyplot as plt


#%% funciones 

def mi_funcion_sen(vmax, dc, ff, ph, nn, fs):
    ts=1/fs
    tt=np.arange(nn)*ts
    xx=dc+vmax*np.sin(2*np.pi*ff*tt+ph)
    return(tt,xx)
# Parámetros

fs = 1000 # Hz
N = 1000 # muestras
vmax=1.5 # (potencial máximo aka apmlitud máxima)
dc=0 # Valor medio (alrededor del que oscila)
ff=1 # frecuencia de la señal
ph=0 # phase

# Sript
tt, xx= mi_funcion_sen(vmax, dc, ff, ph, N, fs)

# Gráfico
plt.figure()
plt.plot(tt,xx) #imprime los dato 
plt.xlabel("Tiempo [s]")
plt.ylabel("Amplitud [V]")
plt.grid(True) #Para mostrar una grilla de fondo
plt.show()
```
